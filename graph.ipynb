```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("C:/Users/acer/Downloads/air_quality_dataset.csv")

# Display dataset
df.head()
```

```python
# Dataset information
print(df.info())

# Statistical description
print(df.describe())

# Check missing values
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x='PM2.5', bins=20, kde=True)

plt.title('Distribution of PM2.5')
plt.xlabel('PM2.5')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df['PM10'])

plt.title('Boxplot of PM10')
plt.ylabel('PM10')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='PM2.5', y='PM10')

plt.title('PM2.5 vs PM10')
plt.xlabel('PM2.5')
plt.ylabel('PM10')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x='CO2', y='Temperature')

plt.title('CO2 vs Temperature')
plt.xlabel('CO2')
plt.ylabel('Temperature')

plt.show()
```
